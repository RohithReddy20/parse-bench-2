# docparse Pilot-150 Perception Runner (Colab A100 — $0 Modal)

Runs all 3 perception pilots sequentially on one A100 (~15 min, ~2.5 Colab compute units), then automatically unassigns the runtime:
1. `exp1_ocria_crops` (238 calls): `llamaindex/OCRia-1` (loop bands, tight table/chart/column crops, cleaned scans)
2. `exp2_qwen3vl_charts` (51 calls): `Qwen/Qwen3-VL-8B-Instruct` + exact vector chart hints
3. `exp3_olmocr2_scans` (65 calls): `allenai/olmOCR-2-7B-1025` on cleaned scanned pages

**Setup:** Upload `pilot150_manifests.tar` and `run_colab_pilot.py` to `MyDrive/docparse/pilot150/` (or directly into `/content/`), choose **Runtime -> Change runtime type -> A100**, and click **Run all**.
